```python
from capture import open_camera, capture_frame
from capture.capture_mac import CameraConfig
from utils import yaml as yml
from utils import paths as p
from pathlib import Path
import cv2
import matplotlib.pyplot as plt

ROOT = p.get_root_dir(Path.cwd())
INPUTS = ROOT / "files" / "inputs"

# load camera parameters
camera_params = yml.load_file(INPUTS / "camera.yaml")
# create camera config object
camera_config = CameraConfig(**camera_params)
# open camera
cap, w, h = open_camera(camera_config)
print(f"actual camera resolution: {w}px x {h}px")
# capture one frame
frame = capture_frame(cap, camera_config)
# display the frame
frame_bgr = cv2.cvtColor(frame, cv2.COLOR_RGB2BGR)
plt.imshow(frame_bgr)

cap.release()
```
