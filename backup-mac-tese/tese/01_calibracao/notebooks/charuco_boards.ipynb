### 1-Create PDF of the charuco pattern to print

```python
from calibration.charuco import charuco_pdf as cpdf
from calibration.charuco import structures as s

from pathlib import Path
from utils import yaml as yml
from utils import paths as p

import cv2

ROOT = p.get_root_dir(Path.cwd())
INPUTS = ROOT / "files" / "inputs"
GENERATED = ROOT / "files" / "generated"

filename = "charuco_7x7_15mm"

# load board info
board_info = yml.load_file(INPUTS / (filename+".yaml"))
print(board_info)

# create charuco board
cb = s.CharucoBoardInfo(*board_info.values())

# create extended charuco board
ecb = s.ExtendedCharucoBoardInfo(cb)

cv2.imwrite(str(GENERATED / (filename+".png")), ecb.img)
print(f"saved {GENERATED / (filename+'.png')}")

ecb.set_center(0.170/2, 0.170/2)
# create pdf with the extended charuco board
cpdf.create_charuco_boards_pdf([ecb], filename = GENERATED / (filename+".pdf"))
```
